# Video Clipper on Colab
Turn a YouTube gameplay stream into vertical clips (face-cam on top, gameplay below, auto-captions).

1. Run the install cell
2. Fill in the config cell
3. Set `PREVIEW = True` first to tune the crop boxes, then `False` to render

```
from pathlib import Path
from IPython.display import Image, display
from clipstack import Settings, Clip, Box, process_all, make_preview

PREVIEW = True   # True = show crop boxes on one frame, False = make clips
SAVE_TO_DRIVE = True

URL = "YouTube-URL"
CLIPS = [
    Clip("00:04:34", "00:05:04", "my_first_clip"),
]
FACECAM = Box(x=.00, y=.40, w=.20, h=.25)
GAME    = Box(x=.21, y=.00, w=.57, h=1.0)

OUT_DIR = "/content/clips"
if SAVE_TO_DRIVE and not PREVIEW:
    from google.colab import drive
    drive.mount("/content/drive")
    OUT_DIR = "/content/drive/My Drive/video Clipper"

s = Settings(url=URL, clips=CLIPS, out_dir=Path(OUT_DIR), facecam=FACECAM, game=GAME,
             cookies=Path("/content/cookies.txt"))

if PREVIEW:
    display(Image(str(make_preview(s))))
else:
    process_all(s)
```
